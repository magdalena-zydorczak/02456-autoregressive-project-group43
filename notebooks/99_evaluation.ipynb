```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

MLP_eval_df = pd.read_csv('../results/eval/generated_sequences_MLP/eval.csv')
unguided_eval_df = pd.read_csv('../results/eval/generated_sequences_rnn_unguided/eval.csv')
# TODO: Add the other models, e.g.:
# classifier_guided_eval_df = pd.read_csv('../results/eval/generated_sequences_rnn_classifier_guided/eval.csv')
# classifier_free_eval_df = pd.read_csv('../results/eval/generated_sequences_rnn_classifier_free/eval.csv')
```

# Unguided generation

```python
fig, ax = plt.subplots(figsize=(8, 6))
parts = ax.violinplot([MLP_eval_df['mean_plddt'], unguided_eval_df['mean_plddt']], showmedians=True, showextrema=False)
for pc, c in zip(parts['bodies'], ['#e8943a', '#2a78d6']):
    pc.set(facecolor=c, edgecolor=c, alpha=0.5)
parts['cmedians'].set_color('black')
ax.set_xticks([1, 2], ['MLP', 'Unguided RNN'])
ax.set_ylabel('Mean pLDDT')
ax.spines[['top', 'right']].set_visible(False)
plt.title('Distribution of Mean pLDDT for Generated Sequences (Unguided generation)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
```

```python
# Lowest and highest scoring sequences for RNN
unguided_eval_df_sorted = unguided_eval_df.sort_values(by='mean_plddt')
unguided_lowest = unguided_eval_df_sorted.iloc[0]
unguided_highest = unguided_eval_df_sorted.iloc[-1]

print("Lowest scoring sequence for RNN:")
print(f"Sample ID: {unguided_lowest['id']}")
print(f"Sequence: {unguided_lowest['sequence']}")
print(f"Mean pLDDT: {unguided_lowest['mean_plddt']}")
print()
print("Highest scoring sequence for RNN:")
print(f"Sample ID: {unguided_highest['id']}")
print(f"Sequence: {unguided_highest['sequence']}")
print(f"Mean pLDDT: {unguided_highest['mean_plddt']}")
```
